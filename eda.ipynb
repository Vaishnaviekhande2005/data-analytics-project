```python
import pandas as pd

df = pd.read_csv(
    r"D:\oasis internship task\Sample - Superstore.csv",
    encoding='latin1'
)

print(df.head())
```

```python
import matplotlib.pyplot as plt

category = df.groupby('Category')['Sales'].sum()

category.plot(kind='bar')

plt.title("Sales by Category")
plt.show()
```

```python
top = df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(10)

print(top)
```

```python
df['Order Date'] = pd.to_datetime(
    df['Order Date'],
    format='mixed'
)

monthly = df.groupby(df['Order Date'].dt.month)['Sales'].sum()

monthly.plot(marker='o')

plt.title("Monthly Sales")
plt.show()
```

```python
import pandas as pd
```

```python
df = pd.read_csv(
    r"D:\oasis internship task\Sample - Superstore.csv",
    encoding='latin1'
)

print(df.head())
```

```python
print(df.shape)
```

```python
import matplotlib.pyplot as plt
import numpy as np

corr = df.corr(numeric_only=True)

plt.figure(figsize=(6,5))
plt.imshow(corr, cmap='coolwarm')
plt.colorbar()

plt.xticks(
    np.arange(len(corr.columns)),
    corr.columns,
    rotation=90
)

plt.yticks(
    np.arange(len(corr.columns)),
    corr.columns
)

plt.title("Correlation Heatmap")
plt.show()
```
